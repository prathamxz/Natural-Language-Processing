# Stemming

#### Stemming is the process of reducing a word to its base or root form (stem) by removing prefixes or suffixes. It is commonly used in linguistics and natural language processing (NLP) to group related words together.

```python
#classification problem
#comments of product is positive review or negative review
#Reviews ------> eating,eat,eaten [going, gone,goes]---->go 

words = ["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalize"]
```

# Porter stemmer

```python
from nltk.stem import PorterStemmer
stemming = PorterStemmer()
```

```python
for word in words :
    print(word +"----->"+ stemming.stem(word))
```

```python
stemming.stem('congratulations')
```

# RegexpStemmer

#### RegexpStemmer is an NLTK class that implements stemming using regular expressions (regex). It takes a regular expression as input and removes the prefix or suffix of a word if it matches the specified pattern.

```python
from nltk.stem import RegexpStemmer
regx = RegexpStemmer('ing$|s$|e$|able$',min=4)
```

```python
regx.stem('eating')
```

```python
regx.stem('congratulates')
```

# Snowball Stemmer
### Snowball Stemmer is a rule-based stemming algorithm in NLTK that removes prefixes and suffixes from words to reduce them to their stem (root form). It is an improved and more flexible version of the Porter Stemmer.
### It is called Snowball because it was developed using the Snowball programming language, which was specifically designed for creating stemming algorithms.

```python
from nltk.stem import SnowballStemmer
Snowball_Stemmer = SnowballStemmer('english')
for word in words :
    print(word+"------>"+Snowball_Stemmer.stem(word))
```

```python
stemming.stem('fairly'),stemming.stem('sportingly')
```

```python
Snowball_Stemmer.stem('fairly'),Snowball_Stemmer.stem('sportingly')
```
