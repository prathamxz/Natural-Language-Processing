# WordNet Lemmatizer

### The **`WordNetLemmatizer`** is a class provided by the **Natural Language Toolkit (NLTK)** library that performs **lemmatization**, a natural language processing (NLP) technique used to reduce words to their base or dictionary form, known as a **lemma**. It uses the **WordNet** lexical database and considers the word's **part of speech (POS)** to produce linguistically correct root words. Unlike stemming, which simply removes prefixes or suffixes, `WordNetLemmatizer` returns meaningful and valid words while preserving their actual linguistic form. For example, it converts **"running"** to **"run"**, **"better"** to **"good"** (when treated as an adjective), and **"mice"** to **"mouse"**.

### **`WordNetLemmatizer` uses the `morphy()` function of `WordNetCorpusReader` to find the correct lemma (base form) of a word using the WordNet lexical database.**

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')  
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
lemmatizer.lemmatize("going")
```

```python
lemmatizer.lemmatize("mouse")
```

```python
"""
POS - 
noun - n
verb - v
adjective - a
adverb - r
"""
lemmatizer.lemmatize("going", pos='v')
```

```python
words = ["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalize"]
```

```python
for word in words:
    print(word+"----->"+lemmatizer.lemmatize(word,pos='v'))
```
